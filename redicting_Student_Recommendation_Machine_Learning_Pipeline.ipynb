Dengan langkah ini, fitur-fitur dataset dikelompokkan berdasarkan jenisnya, memudahkan analisis data lebih lanjut yang memerlukan perlakuan khusus tergantung pada tipe data fitur tersebut.

```python
# list down features based on type
numeric_feature_names = ['ResearchScore', 'ProjectScore']
categoricial_feature_names = ['OverallGrade', 'Obedient']
```

## Penskalaan Fitur Numerik

Selanjutnya gunakan skalar standar dari scikit-learn untuk menormalkan dua atribut berbasis skor numerik dengan menggunakan kode berikut.

Syntax Python di bawah mengimplementasikan normalisasi atau penskalaan standar pada fitur-fitur numerik menggunakan modul `StandardScaler` dari pustaka scikit-learn. Pertama, kita membuat objek `ss` dari kelas `StandardScaler`. Kemudian, objek scaler tersebut dipasang (fit) pada fitur-fitur numerik menggunakan metode `fit`, dengan menggunakan hanya fitur-fitur numerik yang telah dipilih sebelumnya. Setelah pemasangan scaler, fitur-fitur numerik tersebut diubah skala (transform) menggunakan metode `transform`, dan hasilnya disimpan kembali ke variabel `training_features`. Terakhir, hasil dari penskalaan tersebut ditampilkan melalui variabel `training_features`, yang berisi fitur-fitur yang telah diperbarui. Dengan melakukan langkah-langkah ini, fitur-fitur numerik dalam dataset telah dinormalisasi atau diskalakan standar sehingga siap digunakan dalam analisis data atau pemodelan selanjutnya.

```python
from sklearn.preprocessing import StandardScaler
ss = StandardScaler()

# fit scaler on numeric features
ss.fit(training_features[numeric_feature_names])

# scale numeric features now
training_features[numeric_feature_names] = ss.transform(training_features[numeric_feature_names])

# view updated featureset
training_features
```

### Sekarang setelah berhasil menskalakan fitur numerik, selanjutnya akan difokuskan pada penanganan fitur kategorikal serta melakukan perbaikan yang diperlukan berdasarkan kode berikut.

Syntax Python di bawah digunakan untuk melakukan one-hot encoding pada kolom-kolom kategorikal dalam dataframe training_features. Kolom-kolom yang diubah adalah kolom-kolom yang terdaftar dalam variabel categorical_feature_names yaitu terdiri dari 'OverallGrade' dan 'Obedient'.

Fungsi pd.get_dummies akan menghasilkan dataframe baru dengan kolom-kolom baru untuk setiap kategori dalam kolom kategorikal. Setiap kolom baru akan berisi nilai 1 jika kategori tersebut ada di baris tersebut, dan 0 jika tidak.

Setelah one-hot encoding selesai, fitur-fitur baru yang dihasilkan akan ditampilkan untuk diperiksa, memungkinkan pengguna untuk memeriksa apakah pengkodean telah dilakukan dengan benar dan memahami bagaimana variabel-variabel baru tersebut telah dimasukkan ke dalam dataset

```python
training_features = pd.get_dummies(training_features, columns=categoricial_feature_names)
# view newly engineering features
training_features
```

Syntax Python di bawah digunakan untuk mendapatkan daftar nama kolom baru yang dihasilkan setelah proses one-hot encoding.

Pertama, program mengambil semua nama kolom dalam dataframe `training_features` dan mengubahnya menjadi set. Kemudian, program mengambil nama-nama kolom numerik dan mengubahnya menjadi set. Lalu, program mencari selisih antara kedua set tersebut. Hasilnya adalah set yang berisi nama-nama kolom yang tidak ada di kolom numerik, yaitu kolom-kolom kategorikal yang telah diubah menjadi one-hot encoding.

Terakhir, program mengubah set tersebut menjadi daftar. Daftar ini berisi nama-nama kolom baru yang dapat digunakan untuk selanjutnya dalam proses machine learning.


```python
# get list of new categorical features
categorical_engineered_features = list(set(training_features.columns) - set(numeric_feature_names))
```

# Modeling

Selanjutnya akan membangun model klasifikasi sederhana (supervised) berdasarkan set fitur kita dengan menggunakan algoritma algoritma regresi logistik. Kode berikut ini menggambarkan bagaimana membangun model yang diawasi.

Syntax Python di bawah digunakan untuk membuat dan melatih model Regresi Logistik menggunakan fitur yang telah dipersiapkan sebelumnya.

Pertama, LogisticRegression diimpor dari sklearn.linear_model untuk membuat model regresi logistik. Kemudian, np diimpor untuk menggunakan library NumPy yang digunakan untuk manipulasi data numerik. Selanjutnya, model regresi logistik diinisialisasi dengan lr = LogisticRegression(). Model kemudian dibuat dan dilatih dengan memanggil metode 'fit'  pada data latih training_features dan label target np.array(outcome_labels['Recommend']). Terakhir, model menampilkan objek model yang telah dilatih, yang menyimpan informasi seperti koefisien untuk setiap fitur (model.coef_) dan intercept (model.intercept_).

Parameter model menunjukkan koefisien untuk setiap fitur dalam data latih.Koefisien menunjukkan seberapa besar pengaruh masing-masing fitur terhadap prediksi model, sementara intercept adalah bias konstan yang ditambahkan ke prediksi.

```python
from sklearn.linear_model import LogisticRegression
import numpy as np

# fit the model
lr = LogisticRegression()
model = lr.fit(training_features, np.array(outcome_labels['Recommend']))
# view model parameters
model
```

Dengan demikian, sekarang terdapat model pembelajaran supervised berdasarkan model regresi logistik dengan
Regularisasi L2, seperti yang dapat di lihat dari parameter pada output sebelumnya.

# Evaluasi Model

Biasanya, evaluasi model dilakukan berdasarkan beberapa dataset penahanan atau validasi yang berbeda dari dataset training untuk mencegah overfitting atau bias pada model. Karena ini adalah contoh pada dataset mainan, mari evaluasi kinerja model pada data training menggunakan syntax berikut.

Syntax Python di bawah digunakan untuk melakukan evaluasi sederhana pada data latih menggunakan model yang telah dilatih sebelumnya. n. Pertama, model digunakan untuk memprediksi label dari data pelatihan menggunakan metode `predict()`, dan label sebenarnya dari data tersebut diambil dari dataframe `outcome_labels['Recommend']`. Setelah itu, kinerja model dievaluasi menggunakan dua metrik: akurasi dan laporan klasifikasi. Metrik akurasi dihitung dengan membandingkan label sebenarnya dengan label yang diprediksi, menggunakan fungsi `accuracy_score()` dari pustaka Scikit-learn. Laporan klasifikasi yang mencakup presisi, recall, dan f1-score juga dihasilkan menggunakan fungsi `classification_report()`. Hasil evaluasi akhirnya dicetak untuk analisis lebih lanjut.

```python
# simple evaluation on training data
pred_labels = model.predict(training_features)
actual_labels = np.array(outcome_labels['Recommend'])

# evaluate model performance
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report

print('Accuracy:', float(accuracy_score(actual_labels, pred_labels))*100, '%')
print('Classification Stats:')
print(classification_report(actual_labels, pred_labels))
```

Dengan demikian, berdasarkan output di atas dapat dilihat berbagai metrik yang telah disebutkan sebelumnya, seperti akurasi, presisi, recall, dan Skor F1 yang menggambarkan kinerja model.

Data baru yang terdiri dari informasi tentang dua individu, yaitu 'Nathan' dan 'Thomas', dengan atribut-atribut seperti nama, nilai keseluruhan, ketaatan, skor riset, dan skor proyek dimasukkan ke dalam DataFrame 'new_data'.

Selanjutnya, kolom-kolom yang relevan dipilih dengan urutan yang diinginkan menggunakan indexing kolom pada DataFrame. Sekarang kita akan melakukan tugas-tugas yang relevan dengan persiapan data-ekstraksi fitur, rekayasa, dan penskalaan-dalam cuplikan syntax berikut.

Dalam syntax python di bawah, langkah-langkah persiapan data dilakukan untuk mempersiapkan fitur-fitur yang akan digunakan dalam proses prediksi. Pertama, fitur-fitur yang diperlukan untuk prediksi dipilih dari dataframe `new_data` dan disimpan dalam variabel `prediction_features`. Selanjutnya, fitur-fitur numerik dinormalisasi menggunakan skalar yang telah diterapkan sebelumnya pada data latih, untuk memastikan konsistensi skala antara data latih dan data prediksi. Setelah itu, variabel kategorikal diubah menjadi representasi biner menggunakan one-hot encoding dengan fungsi `get_dummies()` dari pustaka Pandas, sehingga dapat digunakan oleh model pembelajaran mesin. Terakhir, dataframe yang berisi fitur-fitur yang telah dipersiapkan tersebut ditampilkan untuk memeriksa kebenaran proses persiapan data.

```python
## data preparation
prediction_features = new_data[feature_names]

# scaling
prediction_features[numeric_feature_names] = scaler.transform(prediction_features[numeric_feature_names])

# engineering categorical variables
prediction_features = pd.get_dummies(prediction_features, columns=categoricial_feature_names)

# view feature set
prediction_features
```

Kita sekarang memiliki fitur yang relevan untuk siswa baru! Namun dapat dilihat bahwa beberapa fitur kategorikal hilang berdasarkan beberapa nilai seperti B, C, dan E. Hal ini karena tidak ada siswa yang memperoleh nilai tersebut, tetapi kita masih membutuhkan atribut tersebut karena model dilatih dengan semua atribut termasuk nilai tersebut. Syntax berikut ini membantu kita mengidentifikasi dan menambahkan fitur kategorikal yang hilang. Kita menambahkan nilai untuk setiap fitur tersebut sebagai 0 untuk setiap siswa karena mereka tidak mendapatkan nilai tersebut.

Syntax Python di bawah bertujuan memastikan kelengkapan fitur kategorikal dalam data baru untuk prediksi.

Prosesnya:

1.   Menentukan Fitur Kategorikal Saat Ini:current_categorical_engineered_features = set(prediction_features.columns) - set(numeric_feature_names) mengidentifikasi fitur kategorikal yang sudah ada dalam prediction_features dengan mengurangkan kolom numerik dari seluruh kolom.
2.   Mengidentifikasi Fitur Kategorikal yang Hilang: missing_features = set(categorical_engineered_features) - current_categorical_engineered_features menentukan fitur kategorikal mana yang seharusnya ada namun belum terdapat dalam prediction_features.
3. Menambahkan Fitur Hilang:
Loop for feature in missing_features:
Menambahkan kolom baru dengan nama feature ke dalam prediction_features.
Dan mengisi kolom tersebut dengan nilai 0 untuk seluruh baris data, karena fitur tersebut tidak ada di data baru.
4. Tampilan Akhir
prediction_features menampilkan DataFrame yang telah diperbarui dengan semua fitur kategorikal yang diperlukan untuk prediksi, termasuk yang sebelumnya tidak ada dalam data baru

Langkah ini penting untuk mencegah error akibat ketidaksesuaian dimensi fitur antara data latih dan data prediksi.





```python
# add missing categorical feature columns
current_categorical_engineered_features = set(prediction_features.columns) - set(numeric_feature_names)
missing_features = set(categorical_engineered_features) - current_categorical_engineered_features
for feature in missing_features:
    # add zeros since feature is absent in these data samples
    prediction_features[feature] = [0] * len(prediction_features)

# view final feature set
prediction_features
```

Kita memiliki serangkaian fitur lengkap yang siap untuk kedua siswa baru. Mari kita uji model kami dan mendapatkan prediksi terkait rekomendasi beasiswa

Syntax Python di bawah, tahapan terakhir dari persiapan data dilakukan untuk memastikan bahwa fitur-fitur yang digunakan untuk melatih model sesuai dengan fitur-fitur yang ada dalam data prediksi.

Dengan proses:


1.   Mencocokan Fitur:
Bagian pertama memastikan bahwa data baru hanya berisi fitur yang digunakan saat melatih model.
prediction_features=prediction_features[training_features.columns] memilih fitur yang relevan dari data baru (prediction_features) berdasarkan fitur yang digunakan untuk melatih model (training_features.columns).
prediction_features.columns=training_features.columns mencocokkan nama kolom pada data baru dengan nama kolom pada data training untuk konsistensi.

2.    Menangani Fitur yang Hilang:
*   Kode memeriksa apakah ada fitur yang ada pada data training tetapi tidak ditemukan pada data baru.
missing_features=set(training_features.columns)-set(prediction_features.columns) membuat daftar fitur yang hilang.
*   Jika ada fitur yang hilang:
Looping iterasi melalui setiap fitur yang hilang (feature).
Kolom baru dengan nama fitur yang hilang ditambahkan ke prediction_features.
Kolom baru diisi dengan nilai 0 (prediction_features[feature]=0) untuk menunjukkan tidak adanya data untuk fitur tersebut.

3. Membuat Prediksi:
Terakhir, kode menggunakan model yang sudah terlatih (model) untuk membuat prediksi pada data baru yang telah disiapkan (prediction_features).
prediction=model.predict(prediction_features) menghasilkan prediksi untuk setiap data baru.

Syntax ini penting untuk memastikan kompatibilitas antara data baru dan data yang digunakan untuk melatih model. Hal ini membantu mencegah kesalahan prediksi dan menghasilkan perkiraan yang lebih akurat.




```python
prediction_features=prediction_features[training_features.columns]
prediction_features.columns=training_features.columns

missing_features=set(training_features.columns)-set(prediction_features.columns)
if missing_features :
  for feature in missing_features :
    prediction_features[feature]=0

    prediction=model.predict(prediction_features)
```

Syntax Python di bawah berfokus pada pembuatan prediksi dan menampilkan hasil.

Penjelasan:


1.   Prediksi:
predictions = model.predict(prediction_features) menggunakan model yang telah dilatih (model) untuk menghasilkan prediksi probabilitas untuk setiap baris data di prediction_features.
